Create a strong code for password authentication using python.

```python
import hashlib
import os

# Function to hash password with salt
def hash_password(password, salt=None):
    if salt is None:
        salt = os.urandom(16)  # Generate random salt
    pwd_hash = hashlib.pbkdf2_hmac('sha256', password.encode(), salt, 100000)
    return salt, pwd_hash

# Function to verify password
def verify_password(stored_salt, stored_hash, entered_password):
    _, new_hash = hash_password(entered_password, stored_salt)
    return new_hash == stored_hash

# Simulated database (in real apps use DB)
user_db = {}

# Register user
def register():
    username = input("Enter username: ")
    password = input("Enter strong password: ")

    salt, pwd_hash = hash_password(password)
    user_db[username] = (salt, pwd_hash)

    print("User registered successfully!\n")

# Login user
def login():
    username = input("Enter username: ")
    
    if username not in user_db:
        print("User not found!\n")
        return

    stored_salt, stored_hash = user_db[username]

    attempts = 3
    while attempts > 0:
        password = input("Enter password: ")
        
        if verify_password(stored_salt, stored_hash, password):
            print("Login successful!\n")
            return
        else:
            attempts -= 1
            print(f"Incorrect password! Attempts left: {attempts}")

    print("Account locked due to too many failed attempts.\n")

# Main program
while True:
    print("1. Register")
    print("2. Login")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == '1':
        register()
    elif choice == '2':
        login()
    elif choice == '3':
        print("Exiting...")
        break
    else:
        print("Invalid choice!\n")
```

```python
# FizzBuzz with counting

def fizzbuzz():
    fizz_count = 0
    buzz_count = 0
    fizzbuzz_count = 0

    for i in range(1, 51):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
            fizzbuzz_count += 1
        elif i % 3 == 0:
            print("Fizz")
            fizz_count += 1
        elif i % 5 == 0:
            print("Buzz")
            buzz_count += 1
        else:
            print(i)

    print("\n--- Counts ---")
    print("Fizz:", fizz_count)
    print("Buzz:", buzz_count)
    print("FizzBuzz:", fizzbuzz_count)

# Run function
fizzbuzz()
```

```python
import time
import numpy as np

# Create data
size = 1000000

# Python list
start = time.time()
py_list = list(range(size))
py_result = [x * 2 for x in py_list]
end = time.time()
print("Python List Time:", end - start)

# NumPy array
start = time.time()
np_array = np.arange(size)
np_result = np_array * 2
end = time.time()
print("NumPy Array Time:", end - start)
```

2. Vectorization improves performance
NumPy applies operations on the entire array at once (vectorization), whereas Python lists process elements one by one using loops.

For large data (like 1 million elements), NumPy is more efficient in both speed and memory usage, making it suitable for scientific and ML tasks.

```python
import pandas as pd

# -------------------------------
# Create Sample Dataset
# -------------------------------
data = {
    "Name": ["Sneha ", "Rahul", "Ananya", "Rahul", None],
    "Marks": [85, 90, None, 90, 75]
}

df = pd.DataFrame(data)

# -------------------------------
# Display original data
# -------------------------------
print("Original Data:\n", df.head())

# -------------------------------
# 1. Handle Missing Values
# -------------------------------
df.fillna(method='ffill', inplace=True)

# -------------------------------
# 2. Remove Duplicates
# -------------------------------
df.drop_duplicates(inplace=True)

# -------------------------------
# 3. Standardize Text
# -------------------------------
df['Name'] = df['Name'].str.strip().str.lower()

# -------------------------------
# Display cleaned data
# -------------------------------
print("\nCleaned Data:\n", df.head())
```

Data cleaning is an essential step in data analysis and machine learning because raw data is often incomplete, inconsistent, and contains errors. In the given dataset, there were missing values, duplicate entries, and inconsistent text formats (like extra spaces and uppercase letters). If such issues are not handled, they can lead to incorrect analysis and poor model performance.

After cleaning, the dataset becomes more accurate and reliable. Missing values were handled using forward fill, duplicates were removed, and text was standardized by converting names to lowercase and removing spaces. This ensures consistency in the data, so that values like “Sneha” and “sneha” are treated as the same entity. Without cleaning, the model might treat them as different, leading to wrong conclusions.

Overall, data cleaning improves the quality, consistency, and usability of the dataset. Assignment (21/03/2026) Assignment Name : Build a Text Cleaner 
Description : Write code to remove punctuation, lowercase text, remove stopwords and test it.

```python
import string
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download once
# import nltk
# nltk.download('punkt')
# nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

def clean_text(text):
    # Lowercase
    text = text.lower()
    
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # Tokenize
    words = word_tokenize(text)
    
    # Remove stopwords
    cleaned = [word for word in words if word not in stop_words]
    
    return " ".join(cleaned)

# Test
sample = "OMG! This movie is sooo good 😍!!!"
print("Original:", sample)
print("Cleaned:", clean_text(sample))
```

Assignment (26/03/2026) Assignment Name : Movie Review Analyzer
Description : Build a simple sentiment analyzer and test on 5 reviews.

```python
def sentiment_analysis(review):
    positive_words = ["amazing", "excellent", "good", "great"]
    negative_words = ["worst", "boring", "bad", "slow"]

    review = review.lower()

    if any(word in review for word in positive_words):
        return "Positive"
    elif any(word in review for word in negative_words):
        return "Negative"
    else:
        return "Neutral"

reviews = [
    "The movie was amazing",
    "Worst movie ever",
    "It was okay",
    "Excellent acting",
    "Very boring film"
]

for r in reviews:
    print(r, "->", sentiment_analysis(r))
```
